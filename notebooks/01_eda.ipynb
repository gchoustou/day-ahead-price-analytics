# 01 — Exploratory Data Analysis

Day-ahead electricity prices — Greek bidding zone (ENTSO-E Transparency Platform).

**Input:** `data/processed/prices_GR.parquet` (hourly; UTC + local time).

## 1. Load & overview

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet('../data/processed/prices_GR.parquet')
df.head()
```

```python
# Summary statistics + missing hours
display(df['price_eur_mwh'].describe())
print('missing hours:', df['price_eur_mwh'].isna().sum())
```

## 2. Time patterns (hour of day / weekday)
Mean price per hour (local time), per weekday, and weekend vs weekday comparison.

```python
by_hour = df.groupby(df['timestamp_local'].dt.hour)['price_eur_mwh'].mean()
by_hour.plot(kind='bar', title='Mean price by hour (local time)');
```

## 3. Monthly / seasonal patterns

```python
monthly = df.groupby(df['timestamp_local'].dt.to_period('M'))['price_eur_mwh'].mean()
monthly.plot(title='Monthly mean price (EUR/MWh)');
```

## 4. Distribution & negative prices
Histogram of prices; count of negative-price hours and when they occur.

```python
df['price_eur_mwh'].hist(bins=80)
neg = df[df['price_eur_mwh'] < 0]
print('negative hours:', len(neg))
neg['timestamp_local'].dt.hour.value_counts().sort_index()
```
